```python
import pandas as pd
data = {
    "name": ["ali", "sara", "ahmed", "ayesha"],
    "age": [20,21,None,22],
    "marks": [85,92,78,None]
}
df = pd.DataFrame(data)
print("original DataFrame")
print(df)
print("\nNull value check")
print(df.isnull())
print(df.isnull().sum())
df["Name"] = df["name"].replace("ali", "ali khan")
print("\nAfter value Replacement")
print(df)
df["marks"] = df["marks"].replace(85,90)
print("\nAfter value Replacement")
print(df)
df = df.drop(2)
print("\nAfter droppong rows")
print(df)
df.drop(1,inplace = True)
print(df)
```
