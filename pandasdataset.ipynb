```python
import pandas as pd
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/a29a0141d20e156043ec257a64c8de3b3a03fd6e/titanic.csv"
df = pd.read_csv(url)
pd.DataFrame(df)

df['age']  = df['age'].fillna(df['age'].mean())
print(df)   #all values in age whuch are missing are filled
```

```python
df['age'].isnull().sum() # with this code we know that age has 0 null value
```

```python
import matplotlib.pyplot as plt
df.plot(x= "age", y = "fare", kind="scatter")
plt.show()
```

```python
import matplotlib.pyplot as plt
df.plot(x= "age", y = "fare", kind="bar")
plt.show()
```

```python
import matplotlib.pyplot as plt
df.plot(x= "age", y = "fare", kind="barh")
plt.show()
```
